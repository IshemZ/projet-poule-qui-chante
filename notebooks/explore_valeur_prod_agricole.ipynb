```python
import pandas as pd
from unidecode import unidecode
```

```python
valeur_prod_agricole = pd.read_csv('../database/valeur_production_agricole_2020.csv', dtype=str)
valeur_prod_agricole = valeur_prod_agricole[
    (valeur_prod_agricole['Code Produit (FAO)'] == '1094') &
    (valeur_prod_agricole['Code Élément'] == '152')
]
valeur_prod_agricole
```

```python
def preprocess_valeur_prod_agricole_data(valeur_prod_agricole):
    valeur_prod_agricole = valeur_prod_agricole.drop(columns=['Description du Symbole', 'Symbole', 'Code année', 'Année', 'Produit', 'Code Produit (FAO)', 'Domaine', 'Code Élément', 'Code Domaine', 'Unité'])
    return valeur_prod_agricole
valeur_prod_agricole = preprocess_valeur_prod_agricole_data(valeur_prod_agricole)
valeur_prod_agricole
```

```python
def convert_valeur_prod_agricole_to_numeric(valeur_prod_agricole):
    valeur_prod_agricole['Valeur'] = valeur_prod_agricole['Valeur'].astype(float)
    return valeur_prod_agricole

valeur_prod_agricole_transformed = convert_valeur_prod_agricole_to_numeric(valeur_prod_agricole)
valeur_prod_agricole_transformed
```

```python
def transform_valeur_prod_agricole_data(valeur_prod_agricole):
    grouper_lignes = valeur_prod_agricole.groupby(['Code zone (FAO)', 'Zone', 'Élément'], as_index=False).sum()
    return grouper_lignes
grouper_lignes = transform_valeur_prod_agricole_data(valeur_prod_agricole_transformed)
grouper_lignes
```

```python
def transform_valeur_prod_agricole_data(grouper_lignes):
    valeur_prod_agricole = grouper_lignes.pivot(index=['Code zone (FAO)','Zone'], columns='Élément', values='Valeur')
    valeur_prod_agricole = valeur_prod_agricole.reset_index()
    valeur_prod_agricole = valeur_prod_agricole.rename(columns={'Code zone (FAO)': 'country_code'})
    return valeur_prod_agricole

valeur_prod_agricole_transformed = transform_valeur_prod_agricole_data(grouper_lignes)
valeur_prod_agricole_transformed
```

```python
def transform_regex_valeur_prod_agricole(valeur_prod_agricole_transformed):
    valeur_prod_agricole_transformed.columns = valeur_prod_agricole_transformed.columns.str.replace(' ', '_', regex=False)
    valeur_prod_agricole_transformed.columns = valeur_prod_agricole_transformed.columns.str.lower()
    valeur_prod_agricole_transformed.columns = [unidecode(col) for col in valeur_prod_agricole_transformed.columns]
    return valeur_prod_agricole_transformed

valeur_prod_agricole_transformed = transform_regex_valeur_prod_agricole(valeur_prod_agricole_transformed)
```

```python
def save_transformed_valeur_prod_agricole_data(valeur_prod_agricole_transformed):
    valeur_prod_agricole_transformed.to_csv('../export/valeur_prod_agricole_2020.csv', index=False, encoding='utf-8')
    print("Valeur production agricole data transformed and saved successfully.")

save_transformed_valeur_prod_agricole_data(valeur_prod_agricole_transformed)
```
