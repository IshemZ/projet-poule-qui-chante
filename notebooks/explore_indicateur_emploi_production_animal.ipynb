```python
indicateur_emploi = pd.read_csv('../database/indicateur_emploi_production_animal_2020.csv', dtype=str)
def preprocess_indicateur_emploi_data(indicateur_emploi):
    indicateur_emploi = indicateur_emploi.drop(columns=['Note', 'Description du Symbole', 'Symbole', 'Code année', 'Source', 'Code source', 'Domaine', 'Code Domaine', 'Unité', 'Code indicateur', 'Indicateur', 'Code Élément', 'Code sexe', 'Sexe'])
    return indicateur_emploi
indicateur_emploi = preprocess_indicateur_emploi_data(indicateur_emploi)
indicateur_emploi
```

```python
def convert_indicateur_emploi_to_numeric(indicateur_emploi):
    indicateur_emploi['Valeur'] = indicateur_emploi['Valeur'].astype(float)
    return indicateur_emploi

indicateur_emploi_transformed = convert_indicateur_emploi_to_numeric(indicateur_emploi)
indicateur_emploi_transformed
```

```python
def transform_indicateur_emploi_data(indicateur_emploi):
    grouper_lignes = indicateur_emploi.groupby(['Code zone (FAO)', 'Zone', 'Élément'], as_index=False).sum()
    return grouper_lignes
grouper_lignes = transform_indicateur_emploi_data(indicateur_emploi_transformed)
grouper_lignes
```

```python
def transform_indicateur_emploi_data(grouper_lignes):
    indicateur_emploi = grouper_lignes.pivot(index=['Code zone (FAO)','Zone'], columns='Élément', values='Valeur')
    indicateur_emploi = indicateur_emploi.reset_index()
    indicateur_emploi = indicateur_emploi.rename(columns={'Code zone (FAO)': 'country_code'})
    return indicateur_emploi

indicateur_emploi_transformed = transform_indicateur_emploi_data(grouper_lignes)
indicateur_emploi_transformed
```

```python
def transform_regex_indicateur_emploi(indicateur_emploi_transformed):
    indicateur_emploi_transformed.columns = indicateur_emploi_transformed.columns.str.replace(' ', '_', regex=False)
    indicateur_emploi_transformed.columns = indicateur_emploi_transformed.columns.str.lower()
    indicateur_emploi_transformed.columns = [unidecode(col) for col in indicateur_emploi_transformed.columns]
    return indicateur_emploi_transformed

indicateur_emploi_transformed = transform_regex_indicateur_emploi(indicateur_emploi_transformed)
```

```python
def save_transformed_indicateur_emploi_data(indicateur_emploi_transformed):
    indicateur_emploi_transformed.to_csv('../export/indicateur_emploi_2020.csv', index=False, encoding='utf-8')
    print("Indicateur emploi data transformed and saved successfully.")

save_transformed_indicateur_emploi_data(indicateur_emploi_transformed)
```
