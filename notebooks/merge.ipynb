```python
# Récupérer les sets de country_code
codes_pop = set(population['country_code'])
codes_sec = set(securite_alimentaire['country_code'])
codes_indices = set(indices_commerciaux['country_code'])
codes_indicateur_emploi = set(indicateur_emploi['country_code'])
codes_valeur_prod_agricole = set(valeur_prod_agricole['country_code'])
codes_culture_prod_poulet = set(culture_prod_poulet['country_code'])
codes_bilan_alimentaire = set(bilan_alimentaire['country_code'])

# country_code présents dans population mais pas dans securite_alimentaire
manquants_dans_secu = codes_pop - codes_sec
print("Présents dans population mais pas dans securite_alimentaire :", manquants_dans_secu)

# country_code présents dans securite_alimentaire mais pas dans population
manquants_dans_pop = codes_sec - codes_pop
print("Présents dans securite_alimentaire mais pas dans population :", manquants_dans_pop)

# country_code présents dans indices_commerciaux mais pas dans population
manquants_dans_pop_indices = codes_pop - codes_indices
print("Présents dans population mais pas dans indices_commerciaux :", manquants_dans_pop_indices)

# country_code présents dans population mais pas dans indices_commerciaux
manquants_dans_indices_pop = codes_indices - codes_pop
print("Présents dans indices_commerciaux mais pas dans population :", manquants_dans_indices_pop)

# country_code présents dans population mais pas dans indicateur_emploi
manquants_dans_pop_indicateur_emploi = codes_pop - codes_indicateur_emploi
print("Présents dans population mais pas dans indicateur_emploi :", manquants_dans_pop_indicateur_emploi)

# country_code présents dans indicateur_emploi mais pas dans population
manquants_dans_indicateur_emploi_pop = codes_indicateur_emploi - codes_pop
print("Présents dans indicateur_emploi mais pas dans population :", manquants_dans_indicateur_emploi_pop)

# country_code présents dans population mais pas dans valeur_prod_agricole
manquants_dans_pop_valeur_prod_agricole = codes_pop - codes_valeur_prod_agricole
print("Présents dans population mais pas dans valeur_prod_agricole :", manquants_dans_pop_valeur_prod_agricole) 

# country_code présents dans valeur_prod_agricole mais pas dans population
manquants_dans_valeur_prod_agricole_pop = codes_valeur_prod_agricole - codes_pop
print("Présents dans valeur_prod_agricole mais pas dans population :", manquants_dans_valeur_prod_agricole_pop) 

# country_code présents dans population mais pas dans culture_prod_poulet
manquants_dans_pop_culture_prod_poulet = codes_pop - codes_culture_prod_poulet
print("Présents dans population mais pas dans culture_prod_poulet :", manquants_dans_pop_culture_prod_poulet)

# country_code présents dans culture_prod_poulet mais pas dans population
manquants_dans_culture_prod_poulet_pop = codes_culture_prod_poulet - codes_pop
print("Présents dans culture_prod_poulet mais pas dans population :", manquants_dans_culture_prod_poulet_pop)

# country_code présents dans population mais pas dans bilan_alimentaire
manquants_dans_pop_bilan_alimentaire = codes_pop - codes_bilan_alimentaire
print("Présents dans population mais pas dans bilan_alimentaire :", manquants_dans_pop_bilan_alimentaire)

# country_code présents dans bilan_alimentaire mais pas dans population
manquants_dans_bilan_alimentaire_pop = codes_bilan_alimentaire - codes_pop
print("Présents dans bilan_alimentaire mais pas dans population :", manquants_dans_bilan_alimentaire_pop)
```

```python
# Codes présents uniquement dans population
uniques_pop = codes_pop - (codes_sec | codes_indices | codes_indicateur_emploi |
                        codes_valeur_prod_agricole | codes_culture_prod_poulet |
                        codes_bilan_alimentaire)

# Afficher les country_code et zone correspondants
print("country_code et zone présents uniquement dans population :")
print(population[population['country_code'].isin(uniques_pop)][['country_code', 'zone']])
```

```python
def merge_all():
    df_merged = pd.merge(groupe_pays, population, left_on='code_pays', right_on='country_code', how='left', suffixes=('_pop', '_secu'))
    df_merged = pd.merge(df_merged, securite_alimentaire, left_on='code_pays', right_on='country_code', how='left', suffixes=('', '_secu'))
    df_merged = pd.merge(df_merged, indices_commerciaux, left_on='code_pays', right_on='country_code', how='left', suffixes=('', '_indices'))
    df_merged = pd.merge(df_merged, indicateur_emploi, left_on='code_pays', right_on='country_code', how='left', suffixes=('', '_emploi'))
    df_merged = pd.merge(df_merged, valeur_prod_agricole, left_on='code_pays', right_on='country_code', how='left', suffixes=('', '_valeur_prod_agricole'))
    df_merged = pd.merge(df_merged, culture_prod_poulet, left_on='code_pays', right_on='country_code', how='left', suffixes=('', '_culture_prod_poulet'))
    df_merged = pd.merge(df_merged, bilan_alimentaire, left_on='code_pays', right_on='country_code', how='left', suffixes=('', '_bilan_alimentaire'))
    return df_merged

df_merged = merge_all()
df_merged
```

```python
def clean_df(df_merged):
    df_merged = df_merged.drop(columns=[
        'country_code_secu', 
        'country_code_indices',
        'country_code_emploi',
        'country_code_valeur_prod_agricole',
        'country_code_culture_prod_poulet',
        'country_code_bilan_alimentaire',
        'zone_secu',
        'zone_indices',
        'zone_emploi',
        'zone_valeur_prod_agricole',
        'zone_culture_prod_poulet',
        'zone_bilan_alimentaire',
        'code_iso3',
        'code_iso2',
        'code_m49',
        'code_groupe_m49',
        'code_pays',
        'zone'
    ])
    df_merged = df_merged.dropna(subset=['country_code'])
    return df_merged

df_merged = clean_df(df_merged)
df_merged
```

```python
def save_merge_csv(df_merged):
    df_merged.to_csv("../export/final_data_2020.csv", index=False, sep=",", encoding="utf-8-sig")
    print("Final data for 2020 transformed and saved successfully.")

save_merge_csv(df_merged)
```
