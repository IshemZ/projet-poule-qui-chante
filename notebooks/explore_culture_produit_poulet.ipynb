```python
import pandas as pd
from unidecode import unidecode
```

```python
culture_prod_poulet = pd.read_csv('../database/culture_produits_poulet_2020.csv', dtype=str)
def preprocess_culture_prod_poulet_data(culture_prod_poulet):
    culture_prod_poulet = culture_prod_poulet.drop(columns=['Note', 'Description du Symbole', 'Symbole', 'Code année', 'Produit', 'Code Produit (FAO)', 'Domaine', 'Code Domaine', 'Unité'])
    return culture_prod_poulet
culture_prod_poulet = preprocess_culture_prod_poulet_data(culture_prod_poulet)
culture_prod_poulet
```

```python
def convert_culture_prod_poulet_to_numeric(culture_prod_poulet):
    culture_prod_poulet['Valeur'] = culture_prod_poulet['Valeur'].astype(float)
    return culture_prod_poulet

culture_prod_poulet_transformed = convert_culture_prod_poulet_to_numeric(culture_prod_poulet)
culture_prod_poulet_transformed
```

```python
def transform_culture_prod_poulet_data(culture_prod_poulet):
    grouper_lignes = culture_prod_poulet.groupby(['Code zone (FAO)', 'Zone', 'Élément'], as_index=False).sum()
    return grouper_lignes
grouper_lignes = transform_culture_prod_poulet_data(culture_prod_poulet_transformed )
grouper_lignes
```

```python
def transform_culture_prod_poulet_data(grouper_lignes):
    culture_prod_poulet = grouper_lignes.pivot(index=['Code zone (FAO)','Zone'], columns='Élément', values='Valeur')
    culture_prod_poulet = culture_prod_poulet.reset_index()
    culture_prod_poulet = culture_prod_poulet.rename(columns={'Code zone (FAO)': 'country_code'})
    return culture_prod_poulet

culture_prod_poulet_transformed = transform_culture_prod_poulet_data(grouper_lignes)
culture_prod_poulet_transformed
```

```python
def transform_regex_culture_prod_poulet(culture_prod_poulet_transformed):
    culture_prod_poulet_transformed.columns = culture_prod_poulet_transformed.columns.str.replace(' ', '_', regex=False)
    culture_prod_poulet_transformed.columns = culture_prod_poulet_transformed.columns.str.lower()
    culture_prod_poulet_transformed.columns = [unidecode(col) for col in culture_prod_poulet_transformed.columns]
    return culture_prod_poulet_transformed

culture_prod_poulet_transformed = transform_regex_culture_prod_poulet(culture_prod_poulet_transformed)
```

```python
def save_transformed_culture_prod_poulet_data(culture_prod_poulet_transformed):
    culture_prod_poulet_transformed.to_csv('../export/culture_prod_poulet_2020.csv', index=False, encoding='utf-8')
    print("Culture production poulet data transformed and saved successfully.")

save_transformed_culture_prod_poulet_data(culture_prod_poulet_transformed)
```
