```python
fao_statistics = pd.read_csv("../export/final_data_2020.csv", index_col="pays")
fao_statistics
```

```python
analyse_univariée = fao_statistics.copy()

analyse_univariée = analyse_univariée.reset_index()  # Remettre 'pays' en colonne
analyse_univariée = analyse_univariée = analyse_univariée.drop(
    columns=[
        'groupe_de_pays',
        'code_groupe_de_pays',
        'country_code',
        'indicateur_emploi',
        'reserves',
        'valeur_des_exportations_-_prix_periode_de_reference',
        'valeur_des_importations_-_prix_periode_de_reference',
        'valeur_des_exportations_-_quantite_periode_de_reference',
        'valeur_des_importations_-_quantite_periode_de_reference',
        'production_brute_2014-2016_(millier_int$)',
        'disponibilite_interieure'
    ],
    errors='ignore'
)

analyse_univariée
```

```python
# Infos générales
analyse_univariée.info()

# Statistiques descriptives
analyse_univariée.describe()

# Valeurs manquantes
missing = analyse_univariée.isnull().sum().sort_values(ascending=False)
missing[missing > 0]
```

```python
from sklearn.impute import SimpleImputer

imputer = SimpleImputer(strategy='median')
colonnes_numeriques = analyse_univariée.select_dtypes(include='number').columns
analyse_univariée[colonnes_numeriques] = imputer.fit_transform(analyse_univariée[colonnes_numeriques])
```

```python
print(analyse_univariée.isna().sum())
```

```python
from sklearn.preprocessing import StandardScaler
X_scaled = StandardScaler().fit_transform(analyse_univariée[colonnes_numeriques])
```

```python
import matplotlib.pyplot as plt
from scipy.cluster.hierarchy import dendrogram, linkage

# linkage_matrix est le résultat de linkage(X_scaled, method='ward')
linkage_matrix = linkage(X_scaled, method='ward')

plt.figure(figsize=(16, 6))
dendrogram(linkage_matrix, labels=analyse_univariée.index, leaf_rotation=90)
plt.axhline(y=100, color='red', linestyle='--')  # <- seuil de coupe
plt.title("Dendrogramme - CAH")
plt.show()

```

```python
from scipy.cluster.hierarchy import fcluster

# Créer les clusters en fixant un seuil
cluster_labels = fcluster(linkage_matrix, t=100, criterion='distance')  # t = hauteur

analyse_univariée["CAH_cluster"] = cluster_labels

cluster_labels
```

```python
cluster_labels = fcluster(linkage_matrix, t=4, criterion='maxclust')  # t = k ici

analyse_univariée["CAH_cluster"] = cluster_labels

cluster_labels
```
