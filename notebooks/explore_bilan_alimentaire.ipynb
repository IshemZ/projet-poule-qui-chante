```python
bilan_alimentaire = pd.read_csv('../database/bilan_alimentaire_2020.csv', dtype=str)
def preprocess_bilan_alimentaire_data(bilan_alimentaire):
    bilan_alimentaire = bilan_alimentaire.drop(columns=['Note', 'Description du Symbole', 'Symbole', 'Code année', 'Produit', 'Code Produit (FAO)', 'Domaine', 'Code Domaine', 'Unité'])
    return bilan_alimentaire
bilan_alimentaire = preprocess_bilan_alimentaire_data(bilan_alimentaire)
bilan_alimentaire
```

```python
def convert_bilan_alimentaire_to_numeric(bilan_alimentaire):
    bilan_alimentaire['Valeur'] = bilan_alimentaire['Valeur'].astype(float)
    return bilan_alimentaire

bilan_alimentaire_transformed = convert_bilan_alimentaire_to_numeric(bilan_alimentaire)
bilan_alimentaire_transformed
```

```python
def transform_bilan_data(bilan_alimentaire):
    grouper_lignes = bilan_alimentaire.groupby(['Code zone (FAO)', 'Zone', 'Élément'], as_index=False).sum()
    return grouper_lignes
grouper_lignes = transform_bilan_data(bilan_alimentaire_transformed)
grouper_lignes
```

```python
def transform_bilan_alim_data(grouper_lignes):
    bilan_alimentaire = grouper_lignes.pivot(index=['Code zone (FAO)','Zone'], columns='Élément', values='Valeur')
    bilan_alimentaire = bilan_alimentaire.reset_index()
    bilan_alimentaire = bilan_alimentaire.rename(columns={'Code zone (FAO)': 'country_code'})
    return bilan_alimentaire

bilan_alimentaire_transformed = transform_bilan_alim_data(grouper_lignes)
bilan_alimentaire_transformed
```

```python
def transform_regex_bilan_alimentaire(bilan_alimentaire_transformed):
    bilan_alimentaire_transformed.columns = bilan_alimentaire_transformed.columns.str.replace(' ', '_', regex=False)
    bilan_alimentaire_transformed.columns = bilan_alimentaire_transformed.columns.str.lower()
    bilan_alimentaire_transformed.columns = [unidecode(col) for col in bilan_alimentaire_transformed.columns]
    return bilan_alimentaire_transformed

bilan_alimentaire_transformed = transform_regex_bilan_alimentaire(bilan_alimentaire_transformed)
```

```python
def save_transformed_bilan_alimentaire_data(bilan_alimentaire_transformed):
    bilan_alimentaire_transformed.to_csv('../export/bilan_alimentaire_2020.csv', index=False, encoding='utf-8')
    print("Bilan alimentaire data transformed and saved successfully.")

save_transformed_bilan_alimentaire_data(bilan_alimentaire_transformed)
```
